# CineMatch Movie Recommendation System

## Step 1: Load Cleaned Dataset
In this step, the cleaned TMDB movie dataset is loaded into pandas for recommendation system development.

```python
import pandas as pd

file_path = r"C:\Users\ay081\Downloads\CineMatch\backend\data\movies_cleaned_final.csv"

df = pd.read_csv(file_path)

print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())

display(
    df[
        ["id", "title", "genres", "keywords", "overview"]
    ].head()
)
```

## Step 2: Inspect Recommendation Features
Here, we verify the columns required for the recommendation model, including genres, keywords, and overview.

```python
required_columns = ["id", "title", "genres", "keywords", "overview"]

print("Required columns available:")
for col in required_columns:
    print(f"{col}: {col in df.columns}")

print("\nSample recommendation features:")
display(df[required_columns].head(10))
```

## Step 3: Handle Missing Values

In this step, missing values in recommendation-related text columns are replaced with empty strings to avoid errors during feature engineering.

```python
text_columns = ["genres", "keywords", "overview"]

print("Missing values before:")
print(df[text_columns].isna().sum())

df[text_columns] = df[text_columns].fillna("")

print("\nMissing values after:")
print(df[text_columns].isna().sum())
```

## Step 4: Create Combined Features

In this step, genres, keywords, and overview are combined into a single text feature for content-based movie recommendation.

```python
df["combined_features"] = (
    df["genres"].astype(str) + " " +
    df["keywords"].astype(str) + " " +
    df["overview"].astype(str)
)

print("Combined feature created successfully.")

display(
    df[["title", "combined_features"]].head(5)
)
```

## Step 5: TF-IDF Vectorization

In this step, the combined movie text features are converted into numerical TF-IDF vectors so that movie similarity can be calculated efficiently.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=30000,
    min_df=2,
    ngram_range=(1, 2)
)

tfidf_matrix = tfidf.fit_transform(df["combined_features"])

print("TF-IDF matrix created successfully.")
print("Matrix shape:", tfidf_matrix.shape)
```

## Step 6: Build Similarity Search

In this step, cosine similarity is calculated only for the selected movie against all other movies. This avoids creating a very large full similarity matrix.

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

df = df.reset_index(drop=True)

title_to_index = pd.Series(
    df.index,
    index=df["title"].str.lower()
).drop_duplicates()

def recommend_movies(movie_title, top_n=10):
    movie_title = movie_title.lower().strip()

    if movie_title not in title_to_index:
        return f"Movie '{movie_title}' not found."

    idx = title_to_index[movie_title]

    similarity_scores = cosine_similarity(
        tfidf_matrix[idx],
        tfidf_matrix
    ).flatten()

    similar_indices = np.argsort(similarity_scores)[::-1][1:top_n + 1]

    recommendations = df.iloc[similar_indices][
        ["id", "title", "genres", "vote_average", "vote_count"]
    ].copy()

    recommendations["similarity_score"] = similarity_scores[similar_indices]

    return recommendations.reset_index(drop=True)

print("Recommendation function fixed successfully.")
```

## Step 6: Rebuild Recommendation Data and TF-IDF Matrix

This step reloads the cleaned dataset and rebuilds the TF-IDF matrix using the same dataframe to ensure index alignment and avoid recommendation lookup errors.

```python
import pandas as pd
import numpy as np

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Load fresh cleaned dataset
file_path = r"C:\Users\ay081\Downloads\CineMatch\backend\data\movies_cleaned_final.csv"

reco_df = pd.read_csv(file_path).reset_index(drop=True)

# Handle missing text
reco_df["genres"] = reco_df["genres"].fillna("")
reco_df["keywords"] = reco_df["keywords"].fillna("")
reco_df["overview"] = reco_df["overview"].fillna("")

# Combine recommendation features
reco_df["combined_features"] = (
    reco_df["genres"].astype(str) + " " +
    reco_df["keywords"].astype(str) + " " +
    reco_df["overview"].astype(str)
)

# Create TF-IDF matrix again from SAME dataframe
tfidf = TfidfVectorizer(
    stop_words="english",
    max_features=30000,
    min_df=2,
    ngram_range=(1, 2)
)

reco_matrix = tfidf.fit_transform(reco_df["combined_features"])

# Create title lookup
title_to_index = pd.Series(
    reco_df.index,
    index=reco_df["title"].str.lower().str.strip()
).drop_duplicates()

print("Movies:", len(reco_df))
print("TF-IDF rows:", reco_matrix.shape[0])
print("Ready:", len(reco_df) == reco_matrix.shape[0])
```

## Step 7: Generate Movie Recommendations

In this step, cosine similarity is calculated between the selected movie and the full movie catalogue to return the most similar titles.

```python
def recommend_movies(movie_title, top_n=10):
    movie_title = movie_title.lower().strip()

    matches = reco_df[
        reco_df["title"].str.lower().str.strip() == movie_title
    ]

    if matches.empty:
        return f"Movie '{movie_title}' not found."

    # If multiple movies have the same title,
    # choose the one with the highest vote count.
    selected_movie = matches.sort_values(
        "vote_count",
        ascending=False
    ).iloc[0]

    idx = selected_movie.name

    similarity_scores = cosine_similarity(
        reco_matrix[idx],
        reco_matrix
    ).flatten()

    similar_indices = np.argsort(similarity_scores)[::-1]

    # Remove the selected movie itself
    similar_indices = similar_indices[
        similar_indices != idx
    ][:top_n]

    recommendations = reco_df.iloc[similar_indices][
        [
            "id",
            "title",
            "release_date",
            "genres",
            "vote_average",
            "vote_count",
            "popularity"
        ]
    ].copy()

    recommendations["similarity_score"] = (
        similarity_scores[similar_indices]
    )

    print("Selected movie:")
    print(
        selected_movie["title"],
        "-",
        selected_movie["release_date"]
    )

    return recommendations.reset_index(drop=True)
```

## Step 8: Validate Recommendation Quality

In this step, recommendations are tested with multiple well-known movies to check whether the model returns relevant and reasonable similar titles.

```python
test_movies = [
    "Avatar",
    "The Dark Knight",
    "Titanic",
    "Interstellar"
]

for movie in test_movies:
    print("\n" + "=" * 60)
    print(f"Recommendations for: {movie}")
    print("=" * 60)

    result = recommend_movies(movie, top_n=5)

    if isinstance(result, str):
        print(result)
    else:
        display(
            result[
                ["title", "release_date", "genres", "vote_average", "similarity_score"]
            ]
        )
```

## Step 9: Save Recommendation Model Files

In this step, the TF-IDF vectorizer, recommendation matrix, and movie index data are saved so they can be reused by the FastAPI backend.

```python
import os
import joblib
from scipy import sparse

model_dir = r"C:\Users\ay081\Downloads\CineMatch\backend\models"
os.makedirs(model_dir, exist_ok=True)

# Save TF-IDF vectorizer
joblib.dump(
    tfidf,
    os.path.join(model_dir, "tfidf_vectorizer.pkl")
)

# Save sparse TF-IDF matrix
sparse.save_npz(
    os.path.join(model_dir, "tfidf_matrix.npz"),
    reco_matrix
)

# Save movie metadata needed for recommendations
reco_df[
    [
        "id",
        "title",
        "release_date",
        "genres",
        "vote_average",
        "vote_count",
        "popularity"
    ]
].to_csv(
    os.path.join(model_dir, "recommendation_movies.csv"),
    index=False
)

print("Recommendation model files saved successfully.")
print("Saved to:", model_dir)
```
